### Problem 1 — Two Sum

Given an array nums and an integer target, return indices of two numbers such that they add up to target.

Example:

nums = [2,7,11,15]
target = 9

Output: [0,1]

```python
nums = [3, 2, 4]
target = 6

# Output: [0,1]


def two_sum(nums, target):
    freq = {}
    last_index = 0

    for index in range(len(nums)):
        required = target-nums[index]
        if required in freq:
            last_index = freq.get(required, 0)
            return [last_index, index]
        else:
            freq[nums[index]] = index

   

print(two_sum(nums, target))
```

Valid Anagram
s = "anagram"
t = "nagaram"

Output = True

Another example:

s = "rat"
t = "car"

Output = False

```python
s = "a"
t = "b"


def valid_anagram(s, t):
    freq = {}

    if (len(s) != len(t)):
        return False

    for chr in s:
        if chr in freq:
            freq[chr] += 1
        else:
            freq[chr] = 1

    for chr in t:
        if chr in freq:
            freq[chr]-=1
            if freq[chr]==0:
                del freq[chr]
                
    if len(freq)==0:
        return True            

    return False

print(valid_anagram(s, t))
```

#### Group Anagrams

Example:

strs = ["eat","tea","tan","ate","nat","bat"]

Output:
[
  ["eat","tea","ate"],
  ["tan","nat"],
  ["bat"]
]

Another example:

strs = [""]

Output:
[[""]]

```python
# first i need 2D array to store all the groups
# then separate array for every group
#strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
strs = [""]


def group_anagram(strs):

    output = None   
    map = {}

    for word in strs:
        sorted_word = "".join(sorted(word))
        if sorted_word in map:
            map[sorted_word].append(word)            
        else:
            map[sorted_word] = [word]
    
    return list(map.values())


print(group_anagram(strs))
```

#### Longest Consecutive Sequence
nums = [100, 4, 200, 1, 3, 2]

Because the longest sequence is:

#### Problem

Given an array of integers, find the first repeating element.

Return the element whose first occurrence index is smallest.

Example

Input:

nums = [10, 5, 3, 4, 3, 5, 6]

Output:

5

Because:

5 repeats
3 also repeats
but 5 appeared earlier first.
Another Example

Input:

nums = [7, 8, 1, 5, 1, 8]

Output:

8

```python
nums = [3, 5, 5, 3]


def first_repeating_element(nums):
    map = {}
    for num in nums:
        if num in map:
            map[num] += 1
        else:
            map[num] = 1

    for key, value in map.items():
        if (value > 1):
            return key

    return -1

print(first_repeating_element(nums))
```

#### Top K Frequent Elements

Given an integer array nums and an integer k, return the k most frequent elements.

Example 1
nums = [1,1,1,2,2,3]
k = 2

Output:

[1,2]
Example 2
nums = [1]
k = 1

Output:

[1]

```python
nums = [1]
k = 1

def top_k_freq(nums, k):
    hashmap = {}
    output = []

    for num in nums:
        if num in hashmap:
            hashmap[num] += 1
        else:
            hashmap[num] = 1

    sorted_hashmap = dict(sorted(hashmap.items(), key=lambda item: item[1],reverse=True))
    print(sorted_hashmap)   

    for key in sorted_hashmap.keys():
        output.append(key)
        if len(output)==k:
            break
            
    return output
          
        


print(top_k_freq(nums, k))
```

#### Next Problem — Ransom Note

Given two strings:

ransomNote
magazine

Return True if you can construct ransomNote using letters from magazine.

Each letter can be used only once.

Example
ransomNote = "aa"
magazine = "aab"

Output:

True

This one is much lighter.

```python
ransomNote = "aa"
magazine = "aab"


def ransom_note(ransomNote, magazine):
    hashmap = {}

    for ch in magazine:
        if ch in hashmap:
            hashmap[ch] += 1
        else:
            hashmap[ch] = 1

    for ch in ransomNote:
        if ch not in hashmap:
            return False
        else:
            hashmap[ch]-=1
            if hashmap[ch]==0:
                del hashmap[ch]
        
    return True

print(ransom_note(ransomNote, magazine))
```

#### Next Greater Element I

Given two arrays nums1 and nums2.

For each element in nums1, find the next greater element in nums2.

If no greater element exists, return -1.

Example

nums2 = [1,3,4,2]

Output:

[-1,3,-1]
nums1 = [2,4]
nums2 = [1,2,3,4]

Output:

[3,-1]

```python
nums1 = [4,1,2] 
nums2 = [1,3,4,2]


def next_greater_element(nums1, nums2):
    stack = []    
    hashmap = {}
    output = []

    for i in range(len(nums2)):

        while stack and nums2[stack[-1]] < nums2[i]:
            index = stack.pop()
            
            hashmap[nums2[index]] = nums2[i]
            print(hashmap)
        stack.append(i)

    for num in nums1:
        if num in hashmap:
            output.append(hashmap[num])
        else:
            output.append(-1)
        print(output)

    return output

print(next_greater_element(nums1, nums2))
```

### Daily Temperatures

Given an array temperatures, return an array such that:

output[i] = number of days until a warmer temperature

If no warmer temperature exists:

0
Example
temperatures = [73,74,75,71,69,72,76,73]

Output:

[1,1,4,2,1,1,0,0]

```python
temperatures = [73, 74, 75, 71, 69, 72, 76, 73]


def daily_temp(temps):
    stack = []
    result = [0] * len(temps)

    for i in range(len(temps)):

        while stack and temps[stack[-1]] < temps[i]:
            index = stack.pop()
            result[index] = i-index
            print(result)
        stack.append(i)

    return result


print(daily_temp(temperatures))
```

### Asteroid Collision

You are given an array of integers asteroids.

Positive value → moving right
Negative value → moving left

When two asteroids collide:

smaller one explodes
if same size → both explode

Return final state after all collisions.

Example 1
asteroids = [5,10,-5]

Output:

[5,10]
Example 2
asteroids = [8,-8]

Output:

[]
Example 3
asteroids = [10,2,-5]

Output:

[10]

```python
asteroids = [4, 7, -3, -8, 5]


def asteroid_collision(asteroids):
    stack = []
    output = []

    for asteroid in asteroids:
        is_alive = True

        while stack and stack[-1] > 0 and asteroid < 0:
            if abs(stack[-1]) < asteroid:
                stack.pop()

            elif abs(stack[-1]) == asteroid:
                stack.pop()
                is_alive = False
                break

            else:
                is_alive = False
                break            

        if is_alive:
            stack.append(asteroid)

    return stack


print(asteroid_collision(asteroids))
```

"abcabccdcdcdef"

Try dry run first, then code.

```python
s = "3[a2[c]]"


def decode_string(s):
    stack = []
    y = ""
    for ch in s:
        x = ""
        if ch == "]":
            while stack and stack[-1] != "[":
                x += stack.pop()

            stack.pop()
            y += x[::-1]*int(stack.pop())

            print(y)
        else:
            stack.append(ch)


print(decode_string(s))
```
