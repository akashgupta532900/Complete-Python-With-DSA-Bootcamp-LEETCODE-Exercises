Problem 1 (Warm-up)

Write a recursive function:

count_vowels("akash") = 2
count_vowels("hello") = 2
count_vowels("xyz") = 0

```python
input = "asshole"


def count_vowles(s):
    if len(s) == 0:
        return 0

    current = s[0]
    if (current == "a" or current == "e" or current == "i" or current == "o" or current == "u"):
        return 1+count_vowles(s[1:])
    else:
        return count_vowles(s[1:])


print(count_vowles(input))
```

```python
s = "abc"


def generate_subseq(s):
    if len(s) == 0:
        return [""]

    current = s[0]

    remaining = s[1:]

    small_ans = generate_subseq(remaining)
    print(small_ans)
    output = []

    for ans in small_ans:
        output.append(current+ans)

    for ans in small_ans:
        output.append(ans)

    return output


print(generate_subseq(s))
```

#### Problem 4 (Phone Keypad).

Without writing code:

For digits "23",

if the recursive call on "3" returns:

["d", "e", "f"]

and current digit "2" maps to:

["a", "b", "c"]

What nested-loop operation do we perform to build the final answer? Explain in one or two lines.

```python
mapping = {
    "2": "abc",
    "3": "def",
    "4": "ghi",
    "5": "jkl",
    "6": "mno",
    "7": "pqrs",
    "8": "tuv",
    "9": "wxyz"
}

input = "7"


def phone_keypad(s):

    if s == "0":
        
        return [""]
    
    current_key=int(s) % 10
    
    current_str = mapping.get(str(current_key))

    small_ans = phone_keypad(str(int(s) // 10))
    print(f"small_ans {small_ans}")
    output = []
    for j in current_str:
        for ans in small_ans:
            output.append(ans+j)
            
    return output

print(phone_keypad(input))
```

Problem 5 — Rat in a Maze

No code.

Given:

maze = [
 [1,1,0],
 [1,1,1],
 [0,1,1]
]

```python
maze = [
    [1, 1, 0],
    [1, 1, 1],
    [0, 1, 1]
]
visited = [[False for _ in range(len(maze))] for _ in range(len(maze))]
print(visited)


def rat_maze(maze, row, col, path):
    n_row = len(maze)
    n_col = len(maze[0])
    if row < 0 or col < 0 or row >= n_row or col >= n_col or visited[row][col] or maze[row][col] == 0:
        return []
    if row == n_row-1 and col == n_col-1:
        return [path]

    

    visited[row][col] = True
    # right
    right = rat_maze(maze, row, col+1, path+[(row, col+1)])

    # left
    left = rat_maze(maze, row, col-1, path+[(row, col-1)])

    # Up
    up = rat_maze(maze, row-1, col, path+[(row-1, col)])

    # down
    down = rat_maze(maze, row+1, col, path+[(row+1, col)])
    
    visited[row][col] = False
    
    return right+left+up+down
    
```

#### Word Search

Given a 2D board of characters and a word, return True if the word exists in the grid, otherwise return False.

Rules:

You can start from any cell.
From a cell, you can move:
Up
Down
Left
Right
You cannot use the same cell more than once in a single path.

Example:

board = [
    ["A","B","C","E"],
    ["S","F","C","S"],
    ["A","D","E","E"]
]

word = "ABCCED"

Output:

True

```python
board = [
    ["A", "B", "C", "E"],
    ["S", "F", "C", "S"],
    ["A", "D", "E", "E"]
]

word = "ABCCKD"

visited = [[False for _ in range(len(board[0]))]for _ in range(len(board))]
print(visited)


def word_search(row, col, index):
    if index == len(word):
        return True

    if row < 0 or col < 0 or row >= len(board) or col >= len(board[0]) or visited[row][col] or board[row][col] != word[index]:
        return False

    visited[row][col] = True

    # right
    right = word_search(row, col+1, index+1)
    # left
    left = word_search(row, col-1, index+1)
    # up
    up = word_search(row-1, col, index+1)
    # down
    down = word_search(row+1, col, index+1)
    visited[row][col] = False
    if up or down or left or right:
        return True
    
    
    return False

print(word_search(0,0,0))
```

### Let’s do N-Queens revision by coding, same style.

Problem:

Place n queens on an n x n board so that no two queens attack each other.

For today use:

n = 4

Expected output should be 2 valid arrangements.

Start with just this part:

board = [["." for _ in range(n)] for _ in range(n)]

```python
n = 4
board = [["." for _ in range(n)] for _ in range(n)]


def is_safe(row, col):
    # upper column
    r = row
    c = col
    while r >= 0:
        if board[r][c] == "Q":
            return False
        r -= 1
    r = row
    c = col
    # upper left diagonal
    while r >= 0 and c >= 0:
        if board[r][c] == "Q":
            return False
        r -= 1
        c -= 1
    r = row
    c = col
    # upper right diagonal
    while r >= 0 and c <= n-1:
        if board[r][c] == "Q":
            return False
        r -= 1
        c += 1
        
    return True   


def n_queens(board, row, path):
    if row == len(board):
        return [path]

    for c in range(len(board[0])):
        if is_safe(row, c):
            board[row][c]="Q"
            output=n_queens(board,row+1,path +[(row,c)]
            board[row][c]="."
            return output
```
