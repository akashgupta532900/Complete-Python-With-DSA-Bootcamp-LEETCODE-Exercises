### Problem: Average of Levels in Binary Tree

Given the root of a binary tree, return the average value of nodes on each level.

Example:

        3
       / \
      9   20
         /  \
        15   7

Output:

[3.0, 14.5, 11.0]

```python
from collections import deque


class TreeNode:
    def __init__(self, val=None, children=None):
        self.val = val
        self.children = children if children else []


root = TreeNode(3)
node9 = TreeNode(9)
node20 = TreeNode(20)
node15 = TreeNode(15)
node7 = TreeNode(7)

root.children.append(node9)
root.children.append(node20)

node20.children.append(node15)
node20.children.append(node7)


def avg_levels(root):
    dq = deque([root])

    output = []

    while dq:
        level_size = len(dq)       
        level_sum = 0
        for i in range(level_size):
            current_node = dq.popleft()
            level_sum += current_node.val

            for eachChild in current_node.children:
                dq.append(eachChild)
                print(dq)

        output.append(level_sum/level_size)
        print(output)
    return output
        
print(avg_levels(root))

```

### Problem — Count Total Nodes in Generic Tree

Given the root of a generic tree, return the total number of nodes present in the tree.

Example:

        1
     /  |  \
    2   3   4
       / \
      5   6

Output:

6

```python
from collections import deque


class TreeNode:
    def __init__(self, val):
        self.val = val
        self.children = []


node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(3)
node4 = TreeNode(4)
node5 = TreeNode(5)
node6 = TreeNode(6)

node1.children.append(node2)
node1.children.append(node3)
node1.children.append(node4)

node3.children.append(node5)
node3.children.append(node6)


def count_Nodes(node1):
    dq = deque([node1])
    count = 0

    while dq:
        dq_size = len(dq)

        for i in range(dq_size):
            current_node = dq.popleft()
            count+=1
            print(count)
            for eachChild in current_node.children:
                dq.append(eachChild)
                
    return count

print(count_Nodes(node1))
    
```
