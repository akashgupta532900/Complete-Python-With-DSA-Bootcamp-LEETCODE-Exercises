Factorial of a number using Recursion
Problem Description:

You are given a non-negative integer n. Your task is to find the factorial of n, defined as the product of all positive integers less than or equal to n. The factorial of 0 is defined to be 1. You must implement this using recursion.



Input:

A non-negative integer n, where 0 <= n <= 20.



Output:

An integer representing the factorial of n.

Example:

Input: n = 5
Output: 120
 
Input: n = 0
Output: 1


```python
n = 20


def fac(n):
    if n == 0:
        return 1
    fact = n*fac(n-1)
    return fact


print(fac(n))
```

Sum of N numbers using Recursion

Input:
A non-negative integer n, where 0 <= n <= 1000.

Output:
An integer representing the sum of the first n natural numbers.

Example:
Input: n = 5
Output: 15  # (1 + 2 + 3 + 4 + 5)
 
Input: n = 0
Output: 0   # Sum of zero natural number

```python
n = 5


def sum_of_natural_numbers(n):
    """
    Function to calculate the sum of the first n natural numbers using recursion.

    Parameters:
    n (int): The non-negative integer for which the sum is to be calculated.

    Returns:
    int: The sum of the first n natural numbers.
    """
    # Your code here
    if (n < 1):
        return 0

    sum = n + sum_of_natural_numbers(n-1)
    return sum

# recurrence relation
# sum_of_natural_numbers(n)=n+sum_of_natural_numbers(n-1)


print(sum_of_natural_numbers(n))
```

Number of Digits using Recursion

Input:
A positive integer n, where 1 <= n <= 10^9.

Output:
An integer representing the number of digits in n.

Example:

Input: n = 12345

Input: n = 7

Fibonacci Series using Recursion

Input:
A non-negative integer n, where 0 <= n <= 30.

Output:
An integer representing the nth Fibonacci number.

Example:
Input: n = 5
Output: 5  # Fibonacci sequence: 0, 1, 1, 2, 3, 5
 
Input: n = 0
Output: 0  # F(0) = 0

```python
n = 5


def Fibonacci_series(n):
    if (n == 0):
        return 0
    elif (n == 1):
        return 1

    smallanswer = Fibonacci_series(n-2)+Fibonacci_series(n-1)

    return smallanswer


print(Fibonacci_series(n))
```

Print N to 1 using Recursion

Input:
A positive integer n, where 1 <= n <= 1000.

Output:
A list of integers from n to 1.

Example:
Input: n = 5
Output: [5, 4, 3, 2, 1]
 
Input: n = 3
Output: [3, 2, 1]


```python
output = []


def count_down(n):
    """
    Function to return a list of integers from n to 1 using recursion.

    Parameters:
    n (int): The positive integer representing the starting point of the range.

    Returns:
    list: A list of integers from n to 1.
    """
    # Your code here

    if (n == 1):
        output.append(1)
        return output
    output.append(n)
    count_down(n-1)
    return output


print(count_down(5))
```

#### Combination Sum

Given an array of distinct positive integers and a target value, return all unique combinations where the chosen numbers sum to the target.

You may use the same number multiple times.

Example
arr = [2,3,5]
target = 8

Output:

[
    [2,2,2,2],
    [2,3,3],
    [3,5]
]
Rules
A number can be chosen multiple times.
[2,2,2,2]

is valid.

Order does not matter.
[2,3,3]

and

[3,2,3]

are considered the same combination.

Return only unique combinations.
If no combination exists, return an empty list.

```python
arr = [2, 3, 5]
target = 8


def Comb_sum(arr, target, index, path):
    if target == 0:
        return [path]
    if target < 0 or index == len(arr):
        return []
    # take 2--> index target-arr[index],path[arr[index]]
    x = Comb_sum(arr, target-arr[index], index, path + [arr[index]])
    # skip 2 --> arr[index+1], target, index+1, path
    y = Comb_sum(arr, target, index+1, path)

    return x+y


print(Comb_sum(arr, target, 0, []))
```

## Sudoku Solver

You are given a partially filled 9 × 9 Sudoku board.

Fill the board so that:

You may assume that:

The input board is valid.
There is exactly one solution.
Example Input
board = [
["5","3",".",".","7",".",".",".","."],
["6",".",".","1","9","5",".",".","."],
[".","9","8",".",".",".",".","6","."],

["8",".",".",".","6",".",".",".","3"],

Expected Output

After solving, the board becomes:

[

```python
board = [
    ["5", "3", ".", ".", "7", ".", ".", ".", "."],
    ["6", ".", ".", "1", "9", "5", ".", ".", "."],
    [".", "9", "8", ".", ".", ".", ".", "6", "."],

    ["8", ".", ".", ".", "6", ".", ".", ".", "3"],
    ["4", ".", ".", "8", ".", "3", ".", ".", "1"],
    ["7", ".", ".", ".", "2", ".", ".", ".", "6"],

    [".", "6", ".", ".", ".", ".", "2", "8", "."],
    [".", ".", ".", "4", "1", "9", ".", ".", "5"],
    [".", ".", ".", ".", "8", ".", ".", "7", "9"]
]
print(len(board))


def find_empty(board):
    for row in range(len(board)):
        for col in range(len(board)):
            if board[row][col] == ".":
                return (row, col)

    return None


def is_safe(row, col, num):

    for c in range(len(board)):
        if board[row][c] == num:
            return False

    for r in range(len(board)):
        if board[r][col] == num:
            return False

    start_of_box_row = (row//3)*3
    start_of_box_col = (col//3)*3

    for box_row in range(start_of_box_row, start_of_box_row+3):
        for box_col in range(start_of_box_col, start_of_box_col+3):

            if board[box_row][box_col] == num:
                return False

    return True


def solved_sudoku(board):
    cell = find_empty(board)
    if cell is  None:
        return True
    
    for num in "123456789":
        if is_safe(cell[0], cell[1], num):
            board[cell[0]][cell[1]] = num
            if solved_sudoku(board):
                return True
            board[cell[0]][cell[1]] = "."
                
    return False

print(solved_sudoku(board))
                
        
```
